```python
import pandas as pd
import plotly.express as px

csv_path = '../../data/processed/NOAA/NOAA_Global_Bleaching_Cleaned.csv'
output_html = '../../outputs/maps/noaa_global_bleaching_map.html'

df = pd.read_csv(csv_path, low_memory=False)

# Drop index-like columns accidentally persisted during CSV export
unnamed_cols = [c for c in df.columns if c.lower().startswith('unnamed')]
if unnamed_cols:
    df = df.drop(columns=unnamed_cols)

df['Latitude_Degrees'] = pd.to_numeric(df['Latitude_Degrees'], errors='coerce')
df['Longitude_Degrees'] = pd.to_numeric(df['Longitude_Degrees'], errors='coerce')
df['Percent_Bleaching'] = pd.to_numeric(df.get('Percent_Bleaching'), errors='coerce')

map_df = df.dropna(subset=['Latitude_Degrees', 'Longitude_Degrees']).copy()

hover_cols = [
    c
    for c in ['Ocean_Name', 'Date', 'Bleaching_Level', 'Percent_Bleaching']
    if c in map_df.columns
]

fig = px.scatter_geo(
    map_df,
    lat='Latitude_Degrees',
    lon='Longitude_Degrees',
    color='Percent_Bleaching',
    color_continuous_scale='Turbo',
    hover_data=hover_cols,
    opacity=0.75,
    title=f'NOAA Global Bleaching Map ({len(map_df):,} points)'
)

fig.update_layout(
    geo={
        'showland': True,
        'landcolor': '#F5F3E7',
        'showocean': True,
        'oceancolor': '#CFE8F3',
        'showcountries': True,
    },
    margin={'l': 0, 'r': 0, 't': 48, 'b': 0},
)
fig.update_traces(marker={'size': 5})

fig.show()
fig.write_html(output_html, include_plotlyjs='cdn')
print(f'Map written to: {output_html}')
```
